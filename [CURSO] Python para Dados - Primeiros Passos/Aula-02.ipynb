5. Crie um programa que solicite dois valores numéricos, um numerador e um denominador, e realize a divisão entre os dois valores. Deixe claro que o valor do denominador não pode ser 0.
6. Crie um programa que solicite dois valores numéricos, um operador e uma potência, e realize a exponenciação entre esses dois valores.
7. Crie um programa que solicite dois valores numéricos, um numerador e um denominador e realize a divisão inteira entre os dois valores. Deixe claro que o valor do denominador não pode ser 0.
8. Crie um programa que solicite dois valores numéricos, um numerador e um denominador, e retorne o resto da divisão entre os dois valores. Deixe claro que o valor do denominador não pode ser 0.

```python
# 1. Crie um programa que solicite dois valores numéricos à pessoa usuária e imprima a soma dos dois valores.
n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
soma = n1 + n2
print(f"A soma de {n1} e {n2} é igual a {soma}.")
```

```python
# 2. Crie um programa que solicite três valores numéricos à pessoa usuária e imprima a soma dos três valores.
n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
n3 = int(input("Digite o terceiro número: "))
soma = n1 + n2 + n3
print(f"A soma de {n1}, {n2} e {n3} é igual a {soma}.")
```

```python
# 3. Crie um programa que solicite dois valores numéricos à pessoa usuária e imprima a subtração do primeiro valor pelo segundo valor.
n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
subtracao = n1 - n2
print(f"A subtração de {n1} e {n2} é igual a {subtracao}.")
```

```python
# 4. Crie um programa que solicite dois valores numéricos à pessoa usuária e imprima a multiplicação dos dois valores.
n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
multiplicacao = n1 * n2
print(f"A multiplicação de {n1} e {n2} é igual a {multiplicacao}.")
```

```python
# 5. Crie um programa que solicite dois valores numéricos, um numerador e um denominador, e realize a divisão entre os dois valores. 
# Deixe claro que o valor do denominador não pode ser 0.
n1 = int(input("Digite o primeiro número: "))
n2 = 0
while n2 == 0:
    n2 = int(input("Digite o segundo número (diferente de 0): "))
divisao = n1 / n2
print(f"A divisão de {n1} e {n2} é igual a {divisao}.")
```

```python
# 6. Crie um programa que solicite dois valores numéricos, um operador e uma potência, e realize a exponenciação entre esses dois valores.
n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
exponenciacao = n1 ** n2
print(f"A exponenciação de {n1} e {n2} é igual a {exponenciacao}.")
```

```python
# 7. Crie um programa que solicite dois valores numéricos, um numerador e um denominador e realize a divisão inteira entre os dois valores. 
# Deixe claro que o valor do denominador não pode ser 0.
n1 = int(input("Digite o primeiro número: "))
n2 = 0
while n2 == 0:
    n2 = int(input("Digite o segundo número (diferente de 0): "))
divisao_int = n1 // n2
print(f"A divisão inteira de {n1} e {n2} é igual a {divisao_int}.")
```

```python
# 8. Crie um programa que solicite dois valores numéricos, um numerador e um denominador, e retorne o resto da divisão entre os dois valores. 
# Deixe claro que o valor do denominador não pode ser 0.
n1 = int(input("Digite o primeiro número: "))
n2 = 0
while n2 == 0:
    n2 = int(input("Digite o segundo número (diferente de 0): "))
resto_div = n1 % n2
print(f"O resto da divisão de {n1} por {n2} é igual a {resto_div}.")
```

```python
# 9. Crie um código que solicita 3 notas de um estudante e imprima a média das notas.
notas = []
for i in range(3):
    while True:
        try:
            notas.append(int(input(f"Digite a {i + 1}ª nota: ")))
            break
        except ValueError:
                print("Valor inválido! Tente novamente:")

media = sum(notas) / len(notas)

print(f"Média das notas: {round(media, 2)}")
```

```python
# 10. Crie um código que calcule e imprima a média ponderada dos números 5, 12, 20 e 15 com pesos respectivamente iguais a 1, 2, 3 e 4.
numeros = [5,12,20,15]
pesos = [1,2,3,4]

soma_ponderada = 0

for i in range(len(numeros)):
    soma_ponderada += numeros[i] * pesos[i]

med_ponderada = soma_ponderada / sum(pesos)

print(f"Média ponderada: {med_ponderada}")
```
